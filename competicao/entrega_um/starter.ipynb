## Sobre este starter

Starter mínimo: carrega os dados, faz uma previsão e salva o arquivo de submissão.
A "previsão" aqui é só um número aleatório para você ver como funciona o fluxo de submissão. Você deve substituir essa parte pelo seu modelo.

Funciona localmente (com seu ambiente python local, lendo de `./data`)
e no Kaggle (lendo de `/kaggle/input/competitions/<COMPETITION_NAME>/`) sem
precisar de nenhuma alteração.

```python
import numpy as np
import pandas as pd
from pathlib import Path

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
```

## Dados

```python
COMPETITION_NAME = 'trab-1-aprendizado-de-maquina-a'
IS_KAGGLE = Path('/kaggle/input').exists()

if IS_KAGGLE:
    INPUT_DIR = Path('/kaggle/input/competitions') / COMPETITION_NAME
else:
    INPUT_DIR = Path('data')

TARGET_NAME = 'loan_status'

train_data = pd.read_csv(INPUT_DIR / 'train.csv')
test_data = pd.read_csv(INPUT_DIR / 'test.csv')

train_data.head()
```

## Fazer a previsão

É aqui que entra o seu modelo. Por enquanto, apenas prevemos um número
aleatório entre 0 e 1 para cada linha

```python
predictions = np.random.uniform(0, 1, size=len(test_data))
```

## Salvar a submissão

```python
submission = test_data[['id']].assign(**{TARGET_NAME: predictions})
submission.to_csv('submission.csv', index=False)
submission.head()
```
